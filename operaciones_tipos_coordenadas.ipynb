## 1. Batería de operaciones aritméticas

Python tiene siete operadores aritméticos. Antes de ejecutar cada celda de esta sección, **prediga** qué cree que va a imprimir — luego ejecute y compare.

| Operador | Nombre | Ejemplo | Da |
|---|---|---|---|
| `+` | suma | `7 + 2` | `9` |

## 2. Comparación y booleanos

Los operadores de comparación **siempre** devuelven un `bool` (`True` o `False`):

`==` (¿son iguales?) · `!=` (¿son distintos?) · `<` `>` `<=` `>=` (menor, mayor, menor o igual, mayor o igual)

**Cuidado:** `=` asigna (semana 1), `==` compara. Son dos operadores distintos con un solo carácter de diferencia — el error más común de quien empieza en Python es escribir uno por el otro.

Para combinar varios booleanos: `and` (los dos deben ser `True`), `or` (al menos uno debe ser `True`), `not` (invierte: `True` pasa a `False` y viceversa).

````python
nivel = 5
tiene_llave = True

# == compara (no asigna). Aqui preguntamos "es nivel igual a 5?"
print("nivel == 5      :", nivel == 5)
print("nivel != 5      :", nivel != 5)
print("nivel > 10       :", nivel > 10)

# and: las dos condiciones deben cumplirse para que el resultado sea True
puede_abrir_cofre = (nivel >= 3) and tiene_llave
print("puede_abrir_cofre:", puede_abrir_cofre)

# or: basta con que UNA condicion se cumpla
puede_entrar_zona = (nivel >= 10) or tiene_llave
print("puede_entrar_zona:", puede_entrar_zona)
````

## 3. Entrada/salida: el perfil de su personaje

Recuerde que`input()` **siempre** devuelve `str`, aunque la persona escriba solo números. Si necesita un número, hay que convertirlo a propósito con `int(...)` o `float(...)`.

Un **f-string** es una forma de construir texto insertando el valor de variables directamente. Se escribe con una `f` antes de las comillas y las variables van entre llaves `{ }`:

```python
nombre = "Ana"
print(f"Hola, {nombre}")   # imprime: Hola, Ana
```

Es más legible que concatenar con `+` (`"Hola, " + nombre`) y evita errores cuando se mezclan tipos (concatenar un `str` con un `int` directamente con `+` da error; con f-strings no).

````python
nombre_personaje = input("Nombre de su personaje: ")

# int(input(...)) hace DOS cosas en una linea: primero pide texto con input(),
# luego convierte ese texto a int con int(...). Si la persona escribe algo
# que no es un numero entero, esto da un error (ValueError) -- lo veremos
# como manejo de errores mas adelante en el curso; por hoy, escriba un numero.
nivel_personaje = int(input("Nivel de su personaje (un numero entero): "))

# f-string: {nombre_personaje} y {nivel_personaje} se reemplazan por su valor real.
print(f"Bienvenido, {nombre_personaje}. Nivel actual: {nivel_personaje}.")

# type() para confirmar: nivel_personaje ya es int, no str, porque lo convertimos.
print("Tipo de nivel_personaje:", type(nivel_personaje))
````

## 4. El mundo de hoy: un mapa con coordenadas

Vamos a trabajar con un mapa sencillo de un juego. Cada posición es una lista de tres números `[x, y, z]` — igual que en la semana 1: `x` izquierda-derecha, `y` abajo-arriba, `z` adelante-atrás.

Representamos el mundo con un **diccionario**: cada llave es el nombre de un objeto y el valor es su posición. Los diccionarios y las listas se estudian a fondo la **semana 3**; hoy los usamos de forma simple, solo para guardar posiciones.

````python
# Posicion del personaje del jugador. Una lista de 3 numeros: [x, y, z].
personaje = [0.0, 0.0, 0.0]

# El mundo: un diccionario. Cada "objeto_del_mundo": [x, y, z].
# {} crea un diccionario vacio; aqui lo llenamos directo con llave: valor.
mundo = {
    "cofre":  [3.0, 0.0, 4.0],
    "portal": [10.0, 0.0, -2.0],
    "npc":    [1.0, 0.0, 1.5],
}

print("Personaje en:", personaje)
for objeto, posicion in mundo.items():
    # .items() entrega pares (llave, valor) del diccionario, uno por uno.
    # El for los va tomando y los guarda en "objeto" y "posicion" en cada iteración.
    print(f"  {objeto} esta en {posicion}")
````

## 5. Calcular distancias (a mano, sin numpy todavía)

La distancia entre dos puntos `(x1, y1, z1)` y `(x2, y2, z2)` en 3D es:

$$d = \sqrt{(x_2-x_1)^2 + (y_2-y_1)^2 + (z_2-z_1)^2}$$

En Python, la raíz cuadrada se puede escribir como `** 0.5` (elevar a la mitad es lo mismo que sacar raíz cuadrada).

````python
def distancia(punto_a, punto_b):
    # punto_a y punto_b son listas [x, y, z]. Los indexamos con [0], [1], [2]
    # para sacar x, y, z de cada uno (el indexado de listas se ve a fondo en la semana 3).
    dx = punto_b[0] - punto_a[0]
    dy = punto_b[1] - punto_a[1]
    dz = punto_b[2] - punto_a[2]
    return (dx**2 + dy**2 + dz**2) ** 0.5   # ** 0.5 es la raiz cuadrada

RADIO_INTERACCION = 5.0   # unidades del mundo; en MAYUSCULAS por convencion, para
                          # marcar que es una constante que no pensamos cambiar

for objeto, posicion in mundo.items():
    d = distancia(personaje, posicion)
    dentro_del_radio = d <= RADIO_INTERACCION   # bool: True si esta lo bastante cerca
    print(f"  distancia a {objeto}: {d:.2f}  ->  dentro del radio de interaccion: {dentro_del_radio}")
````

## 6. El problema: mover al personaje (aquí las listas engañan)

Queremos **mover** al personaje: sumarle un **desplazamiento** a su posición actual. Como el personaje y el desplazamiento son listas de 3 números, parece razonable sumarlas con `+`, igual que se suman dos números. Probemos.

````python
desplazamiento = [1.0, 0.0, 2.0]   # "muevase 1 en x, 0 en y, 2 en z"

nueva_posicion_preliminar = personaje + desplazamiento
print("Posicion despues de 'mover' (version ingenua):", nueva_posicion_preliminar)
print("Cuantos numeros tiene esa 'posicion'?:", len(nueva_posicion_preliminar))
````

**Mire el resultado.** No hubo ningún error rojo. Pero una posición en 3D tiene **tres** números, `[x, y, z]` — y lo que imprimió arriba tiene **seis**. `personaje + desplazamiento` no sumó componente a componente: **pegó una lista detrás de la otra**. En Python, `+` entre dos listas significa "concatenar" (unir), no "sumar como vector" — así se vean parecidas a un punto en el espacio.

Es el mismo tipo de problema que la rotación en grados de la semana 1: el código **corrió sin protestar**, pero el resultado es una posición sin sentido para nuestro mundo. `Python` nos nos avisó.

## 7. La corrección — y numpy, de refilón

Hay dos formas de arreglarlo hoy:

**(a) A mano, componente a componente**, con una lista nueva construida a partir de las dos anteriores:

````python
# Sumamos x con x, y con y, z con z -- por posicion (indice 0 con 0, 1 con 1, 2 con 2).
nueva_posicion_manual = [
    personaje[0] + desplazamiento[0],
    personaje[1] + desplazamiento[1],
    personaje[2] + desplazamiento[2],
]
print("Posicion despues de moverse (a mano, componente a componente):", nueva_posicion_manual)
````

**(b) Con `numpy`.** `numpy` es una librería para trabajar con números y vectores (la usó de refilón en la primera escena de la semana 1). Su tipo especial, `np.array`, **sí** interpreta `+` como suma componente a componente cuando ambos lados son arrays — no como concatenar.

````python
import numpy as np

# np.array(...) convierte una lista comun de Python en un "array" de numpy.
# Se ve casi igual al escribirla, pero se comporta distinto con los operadores.
personaje_np = np.array(personaje)
desplazamiento_np = np.array(desplazamiento)

# Ahora si: "+" entre dos arrays de numpy SUMA componente a componente.
nueva_posicion_numpy = personaje_np + desplazamiento_np
print("Posicion despues de moverse (con numpy):", nueva_posicion_numpy)

# Comprobacion: deberia coincidir con la version manual de arriba.
print("Coincide con el calculo a mano?:", list(nueva_posicion_numpy) == nueva_posicion_manual)
````
