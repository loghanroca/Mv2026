## 1. El editor: una función con una **prueba**

Hasta ahora usted ha construido un escenario empleando diccionarios definidos a mano. Esa opción no permite construir escenarios a una mayor escala: un mundo virtual se obtiene a partir de datos crudos (un archivo, un formulario) y hay que **construir** cada objeto verificando que tenga sentido. Esa lógica — con validación interna — es exactamente el tipo de código que **conviene escribir en un archivo, no en una celda**: se reutiliza entre cuadernos y no depende del orden en que ejecutó las celdas.

** Abra `galeria_utils.py` en su editor. Ya tiene una función:

```python
def crear_obra(nombre, artista, posicion, sala, año, visible=True):
    assert isinstance(nombre, str) and nombre, "el nombre debe ser un texto no vacio"
    assert isinstance(artista, str) and artista, "el artista debe ser un texto no vacio"
    assert len(posicion) == 3, "la posicion debe tener exactamente 3 coordenadas (x, y, z)"
    assert isinstance(año, int), "el año debe ser un numero entero"
    return {
        "nombre": nombre, "artista": artista, "posicion": posicion,
        "sala": sala, "año": año, "visible": visible,
    }
```

Cada `assert` es un **prueba**: una condición que debe cumplirse para que la función se ejecute. Si no
se cumple, Python lanza `AssertionError` con el mensaje que usted escribió — en vez de construir un
objeto a medias y descubrir el problema tres celdas después.

````python
from galeria_utils import crear_obra

# Un escenario válido: los cuatro asserts se cumplen, la funcion devuelve el diccionario.
obra_de_prueba = crear_obra("Prueba", "Anónimo", (1.0, 2.0, 3.0), "Sala Norte", 2024)
print(obra_de_prueba)
````

Una prueba que falla a propósito. Esto es un error esperado: `crear_obra` está haciendo exactamente su trabajo al rechazar un dato que no tiene sentido.

````python
try:
    crear_obra("Prueba mala", "Anónimo", (1.0, 2.0), "Sala Norte", 2024)  # posicion con 2 numeros, no 3
except AssertionError as e:
    print("AssertionError, como se esperaba:", e)
````

## 3. Construir la escena sin que los datos errados se pierdan en silecio

**`[Trabajo del estudiante — # TODO 1]`** Recorra `datos_crudos` con un `for`. Para cada tupla:

1. Desempaquétela en seis variables (`nombre, artista, posicion, sala, año, visible = dato`).
2. Dentro de un bloque `try`, llame a `crear_obra(...)` con esas variables y agregue el resultado a
   una lista `galeria`.
3. Si `crear_obra` lanza `AssertionError`, captúrela con `except AssertionError as e` y agregue una
   tupla `(nombre, str(e))` a una lista `errores` — **no** deje que el error detenga el ciclo entero
   por un solo dato malo.

Al final, imprima cuántas obras quedaron válidas en `galeria` y muestre el contenido de `errores`.
Esto es el mismo criterio que se ha empleado antes: un dato que no sirve **se documenta**, no desaparece sin dejar rastro.

````python
# TODO 1: construya galeria y errores recorriendo datos_crudos con try/except.
galeria = []
errores = []

for dato in datos_crudos:
    nombre, artista, posicion, sala, año, visible = dato
    try:
        obra = crear_obra(nombre, artista, posicion, sala, año, visible)
        galeria.append(obra)
    except AssertionError as e:
        errores.append((nombre, str(e)))

print(f"Obras válidas en galería: {len(galeria)}")
print("Lista de errores capturados:")
for error in errores:
    print(error)
````

## 4. El choque: una función que no es tan independiente como parece

Ahora una función para crear una **sala** vacía, lista para recibir obras.

````python
def crear_sala(nombre, obras=[]):
    return {"nombre": nombre, "obras": obras}

sala_norte = crear_sala("Sala Norte")
sala_sur = crear_sala("Sala Sur")

sala_norte["obras"].append("El Reflejo Digital")

print("sala_norte:", sala_norte)
print("sala_sur:  ", sala_sur)
````

**Mire `sala_sur`.** Nunca le agregó nada — y sin embargo tiene `"El Reflejo Digital"`
adentro. **No hubo ningún error.** `crear_sala("Sala Sur")` se ve como una llamada independiente de
`crear_sala("Sala Norte")`, pero las dos terminaron compartiendo **la misma lista**.

**¿Por qué?** El valor por defecto `obras=[]` se crea **una sola vez**, cuando Python lee la
definición de la función — no cada vez que la llama. Todas las llamadas que no pasan su propio
`obras` reciben **la misma lista**, guardada en memoria desde que se definió `crear_sala`. Es el
"no da error pero está mal" de esta semana: no es un valor centinela ni un tipo mezclado, se trata de un **estado compartido por accidente** entre llamadas que parecían no tener nada que ver entre sí.

````python
print("¿'sala_norte[\"obras\"]' y 'sala_sur[\"obras\"]' son la MISMA lista de Python?")
print(sala_norte["obras"] is sala_sur["obras"])
````

## 5. La corrección: `None` como valor por defecto

La solución estándar en Python: el valor por defecto es `None` (un valor fijo, no un objeto que se
podría compartir), y **dentro** de la función, si no llegó nada, ahí sí se crea una lista **nueva**
en cada llamada.

— ya está en `galeria_utils.py`:

```python
def crear_sala_segura(nombre, obras=None):
    if obras is None:
        obras = []
    return {"nombre": nombre, "obras": obras}
```

````python
from galeria_utils import crear_sala_segura

sn2 = crear_sala_segura("Sala Norte")
ss2 = crear_sala_segura("Sala Sur")
sn2["obras"].append("El Reflejo Digital")

print("sn2:", sn2)
print("ss2:", ss2)
print("¿misma lista esta vez?", sn2["obras"] is ss2["obras"])
````
