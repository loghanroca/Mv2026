## 1. ¿Qué es una celda?

Un cuaderno de Jupyter (`.ipynb`) es una secuencia de **celdas**. Cada celda es de uno de dos tipos:

- **Celda de markdown** (como esta): texto con formato. No es código Python; solo se renderiza como texto cuando la ejecuta.
- **Celda de código**: Python de verdad. Cuando la ejecuta, el código **corre** como un programa y el resultado aparece justo debajo.

Ejecute la celda de código de abajo con `Shift+Enter` (o el botón ▶ en la barra de Jupyter).

````python
# Esto es una celda de codigo. Todo lo que empieza con "#" es un COMENTARIO:
# Python lo ignora por completo; esta ahi para que un humano lo lea.

# print(...) es una FUNCION: un bloque de codigo con nombre que hace algo cuando lo "llama".
# Aqui la llamamos con un texto entre comillas -eso es un str, lo vera en la seccion 2-
# y ella lo muestra en pantalla, justo debajo de la celda.
print("Si ve este mensaje debajo de la celda, la acaba de ejecutar correctamente.")
````

## 2. El kernel: el "cerebro" que recuerda

Cuando ejecuta una celda de código, no se ejecuta sola: hay un proceso de Python corriendo detrás, llamado **kernel**, que **recuerda** todo lo que ha ejecutado hasta ese momento — variables, funciones, lo que sea.

Fíjese en el número entre corchetes que aparece a la izquierda de una celda de código después de ejecutarla, por ejemplo `In [1]:`. **Ese número es el orden real de ejecución** — no el orden en que la celda aparece en la página. Si usted ejecuta las celdas en el orden en que están escritas, ambos órdenes coinciden. Pero si vuelve atrás y ejecuta una celda de nuevo, o salta una, **dejan de coincidir** — y eso es exactamente lo que vamos a provocar en la sección 4.

Por ahora, quédese con la idea: **el kernel tiene memoria y esa memoria depende de lo que usted ejecutó, en qué orden, pero no de lo que ve escrito en la pantalla.**

## 3. Variables y los cuatro tipos básicos

Una **variable** es una etiqueta con un valor pegado. En Python se crea con el signo `=`:

```python
edad = 20
```

**Cuidado:** en Python, `=` **no** significa "es igual a" como en una ecuación de matemáticas. Significa **"asigna"**: toma el valor de la derecha (`20`) y pégalo a la etiqueta de la izquierda (`edad`). Para comparar si dos cosas son iguales, Python usa `==` (dos signos) — lo verá el miércoles.

Python tiene muchos tipos de datos, pero hoy nos interesan cuatro:

| Tipo | Qué guarda | Ejemplo |
|---|---|---|
| `int` | números enteros, sin punto decimal | `5`, `-3`, `1000` |
| `float` | números con punto decimal ("flotante": el punto puede estar en cualquier posición) | `3.14`, `-0.5`, `2.0` |
| `str` | texto, siempre entre comillas simples o dobles | `"hola"`, `'mv2026'` |
| `bool` | solo dos valores posibles: `True` o `False` (con mayúscula inicial, así los reconoce Python) | `True`, `False` |

Ejecute la celda siguiente y observe cómo `type()` —otra función, igual que `print()`— le dice a qué tipo pertenece cada variable.

````python
# Cuatro variables, una de cada tipo. Fijese que NINGUNA linea dice de que tipo es:
# Python lo DEDUCE de como escribio el valor (esto se llama "tipado dinamico").
edad = 20              # sin punto decimal -> Python lo guarda como int
altura_m = 1.72         # con punto decimal -> float
nombre = "Valentina"    # entre comillas -> str
esta_matriculada = True # True/False, sin comillas -> bool

# type(variable) devuelve el tipo. Lo imprimimos junto con el valor para verlo claro.
# La coma dentro de print() separa varios argumentos: print() los muestra
# uno despues del otro, separados por un espacio, en la misma linea.
print("edad:", edad, "->", type(edad))
print("altura_m:", altura_m, "->", type(altura_m))
print("nombre:", nombre, "->", type(nombre))
print("esta_matriculada:", esta_matriculada, "->", type(esta_matriculada))
````

## 4. `print()` no es lo único que muestra algo

Hay una diferencia importante que confunde a casi todo el mundo al empezar:

- `print(x)` **siempre** muestra `x` en pantalla, sin importar en qué parte de la celda esté.
- Si en la **última línea** de una celda escribe solo el nombre de una variable (sin `print()`), Jupyter también la muestra — pero **solo si es la última línea**. Si es cualquier otra línea, no se muestra nada.

Ejecute la celda de abajo y observe cuál de las dos líneas se muestra.

````python
numero_favorito = 7
numero_favorito          # esta linea NO es la ultima de la celda -> no se muestra nada
numero_favorito * 2      # esta SI es la ultima linea -> Jupyter la muestra sola, sin pedirselo
# Si quiere estar seguro de que algo se muestra, no confie en "es la ultima linea":
# use print() explicitamente, como hicimos arriba. Es mas claro y no depende de la posicion.
````

## 5. `input()`: pedirle algo a quien usa el programa

**El detalle que hay que recordar siempre:** todo lo que devuelve `input()` es **texto (`str`)**, así la persona haya escrito solo números.
